```python
from crawler import Beatport, Track, Artist, Label
from pprint import pprint
from IPython.display import display, Markdown
from typing import List
from toolz import unique, pipe, groupby
```

```python
b = Beatport()
```

```python
def display_track(track: Track):
    audio = f'<audio width="320" height="240" src="{track.sample}" controls> </audio>'
    image = f'<img src={track.image} width="48">'
    display(Markdown(f'<div>{image} {audio} <br/> {track} <br/> {track.release_date}</div>'))
    
```

```python
def display_artist(artist: Artist):
    display(Markdown(f'<img src={artist.image} width="128"> <br/> {artist}'))
```

```python
def display_label(label: Label):
    display(Markdown(f'<img src={label.image} width="128"> <br/> {label}'))
```

```python
def all_artist_labels_by_date(tracks: List[Track]):
    s = sorted(tracks, key=lambda x:x.release_date)
    g = groupby(lambda x: x.label.name, s)
    return g
```

```python
ar, _ = b.search('CHRIZ (GER)')
ari = ar[0]
ari.enrich(b)
```

```python
d = all_artist_labels_by_date(ari.tracks)
for k in d:
    display_label(d[k][0].label)
print()
for k in d:
    display_label(d[k][0].label)
    for track in d[k]:
        display_track(track)
```
